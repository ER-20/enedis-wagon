```python
#En te basant sur le dictionnaire précédent, fais moi le code qui me permettra de faire un groupby sur 'Identifiant__BAN'
#- je veux agréger toutes les colonnes numériques en faisant une moyenne
#- je veux agréger toutes les colonnes textuelles en prenant la valeur la plus présente (et en choisissant arbitrairement lorsque plusieurs valeurs sont présentes le même nombre de fois)
```

```python
# Separate the numerical and textual columns
num_cols = df_filtered_2.select_dtypes(include=['float64', 'int64']).columns
text_cols = df_filtered_2.select_dtypes(include=['object']).columns
```

```python
agg_funcs = {}
for col in num_cols:
    agg_funcs[col] = 'mean'
for col in text_cols:
    agg_funcs[col] = lambda x: x.value_counts().index[0] if len(x.value_counts()) > 0 else pd.NA

result = df_filtered_2.groupby("Identifiant__BAN").agg(agg_funcs)
```
